# DEL A — Projektet som sammenhængende teknisk fortælling

## 5. Appen: Hvorfor Streamlit og fire trin?

**Faktisk design:** `app.py` styrer flowet **Upload → Role → Check → Report**. Det holder UI-logikken adskilt fra modulerne til CV-parsing, matching, demand og rapportering. `skillradar/ui.py` laver genbrugelige HTML/CSS-komponenter. `st.session_state` holder brugerens CV-tekst, rolle, dataset og manuelle rettelser gennem appens reruns. `st.cache_resource` genbruger store, ikke-private ressourcer som matcher og posting-tabeller.

**Hvorfor?** Streamlit gør det muligt at bygge en webprototype i Python uden et separat JavaScript-framework. Fire trin begrænser den information, brugeren skal tage stilling til ad gangen. Check-trinnet håndterer kendte matcherfejl, inden rapporten produceres.

**Kritisk spørgsmål:** Hvorfor ikke Flask/React? **Svar:** Det kunne give mere kontrol over routing og komponenter, men ville forøge frontend-arbejdet uden tydelig forskningsgevinst i en prototype.

**Vigtigt:** CV'et bliver behandlet i sessionens hukommelse i `app.py`; man bør ikke påstå, at det er anonymiseret eller forsvinder straks efter hvert klik. LLM-forsøg på eksterne tjenester skal omtales særskilt fra webappens lokale keyword-flow.

## 6. CV-parsing: Fra fil til ren tekst

`skillradar/cv_parser.py` accepterer `.pdf`, `.docx`, `.txt` og `.md`. PDF læses med `pdfplumber`, DOCX med `python-docx` (også tabeller), og plain text dekodes som UTF-8. `clean_text` håndterer blandt andet typografiske ligaturer, specialtegn, orddeling over linjer og whitespace. Hvis der er under 50 tegn efter rensning, giver parseren en fejl, fordi en scannet PDF uden tekstlag ikke kan behandles pålideligt.

**Metodevalg:** I valgte en simpel deterministisk parser, så et CV kan omsættes til tekst. **Alternativ:** OCR på scannede PDF'er. Det ville udvide dækningen, men giver flere fejl, ekstra tid og højere kompleksitet.

**Kritisk pointe:** I kan kun identificere værktøjer, der overlever filudtrækket og faktisk står i teksten.

## 7. Jobdatakilder: Hvorfor to?

Primary har ca. 1,3 mio. LinkedIn-opslag i hele datakilden fra januar 2024 og en stor tekstfil (`job_summary.csv`). Secondary har ca. 124.000 rå opslag fra april 2024, er USA-fokuseret og giver entry-level/internship-information. Begge er snapshots.

**Formål med primary:** større antal relevante opslag. **Formål med secondary:** replikation på en uafhængig kilde og entry-level-vinklen.

**Kritisk pointe:** I sammenligner ikke automatisk lige populationer; kildernes skema, dækning og scrapers kan være forskellige. Derfor er det ikke forsvarligt at kalde forskellen mellem januar og april for en trend.

## 8. O*NET og den præcise definition af skills

O*NET's `software_skills.csv` bruges til en kanonisk liste af navngivne teknologier, som er markeret `Hot Technology` **eller** `In Demand` og som ikke frasorteres som for generiske. Målene er konkrete værktøjer som Python, SQL eller Tableau — ikke brede kompetencer som machine learning, statistik eller kommunikation.

**Hvorfor fælles ordforråd?** Samme teknologi kan stå på flere måder i CV og jobannonce; fælles repræsentation gør sammenligning mulig.

**Risiko:** O*NET dækker ikke alle relevante teknologier, især nyere produkter. Det er en *vocabulary ceiling*, som jeres test med out-of-vocabulary labels tydeliggør.

## 9. Keyword-matcher: Hvad gør den konkret?

`skillradar/matcher.py` bruger tokens, n-grams og en `alias_to_tool`-ordbog. Lange O*NET-navne suppleres med kortere aliaser; forkortelser dannes kun under bestemte regler. Nogle tvetydige navne matches med case sensitivity; `MANUAL_DROP` fjerner aliaserne `Google` og `Facebook` efter manuel fejlundersøgelse. `extract_tools(text)` returnerer et **set** af kanoniske værktøjer; `evidence(text)` giver tekststykker til brugeren.

**Hvorfor?** Det er hurtigt, transparent og kan inspiceres. **Problemer:** tvetydige ord, kontekstfejl og værktøjer uden for vokabularet.

**Hvis du bliver spurgt »er det machine learning?«:** Selve keyword-matcheren er regelbaseret, ikke trænet. LLM-metoderne blev afprøvet i en separat sammenligning.

## 10. Offline preprocessing: Hvorfor bygger I tabeller?

`skillradar/demand.py` vælger relevante jobtitler med regex, bruger matcher på posting-tekst og gemmer én række pr. posting med metadata, `text_hash` og et værktøjssæt. Primary-teksten streames i chunks. `load_posting_tools()` genskaber værktøjssæt og fjerner opslag med samme normaliserede teksthash **inden for samme rolle**.

**Begrundelse:** Hurtig app, lavere hukommelsesforbrug, færre gentagne opslag og fælles værktøjsrepræsentation.

**Nuance:** Hashededuplikering finder identisk tekst efter whitespace/case-normalisering; den er ikke en fuld near-duplicate-algoritme. Hvis matcherreglerne ændres, skal posting-tabellerne rebuildes for at holde CV-siden og posting-siden konsistente.

## 11. Demand, Wilson og coverage: Hvordan beregner I rapporten?

**Demand:** \(p_t = k_t/n\), hvor \(k_t\) er antal opslag for rollen, der nævner teknologi \(t\) mindst én gang, og \(n\) er antal relevante opslag. `demand.py` beregner Wilson 95 %-intervaller for disse andele.

**Coverage:** `report.py` filtrerer til værktøjer med mindst 5 % share, tager op til top-15 og beregner \(\sum_{t\in CV\cap Top}p_t / \sum_{t\in Top}p_t\). Det er en *demand-vægtet dækningsandel af topværktøjerne*, ikke en præcis sandsynlighed for at blive ansat.

**Hvorfor vægte?** At mangle et værktøj i 70 % af opslagene skal tælle mere end et værktøj i 7 %.

**Begrænsning:** Det siger ikke, om værktøjet er obligatorisk i hver annonce, eller hvor stærk brugerens kompetence er.

## 12. Robusthed: Spearman, bootstrap og senioritet

`skillradar/robustness.py` tester tre ting: (a) replication med Spearman-korrelation mellem rangeringer på tværs af datakilder, (b) bootstrap-rankintervaller inden for en kilde, (c) junior/senior-forskelle med to-proportioners z-test og Benjamini–Hochberg-korrektion.

**Hvorfor Spearman?** I undersøger først og fremmest, om *rækkefølgen* af værktøjer ligner hinanden, ikke om de absolutte andele er identiske.

**Hvorfor bootstrap?** For at se hvor meget rangeringen kunne flytte sig ved resampling af opslag.

**Hvorfor BH?** Når mange værktøjer testes, vil enkelte lave p-værdier forekomme ved tilfældighed. Korrigerede q-værdier reducerer risikoen for falske fund.

**Begrænsning:** Replikation på to scrape-kilder er ikke bevis for repræsentativitet for alle jobs.

## 13. LLM-only, Hybrid og HybridVerify: Hvad afprøvede I?

`skillradar/llm_matcher.py` implementerer tre sammenlignelige interfaces (`extract_tools(text) -> set[str]`):
- **LLMOnly:** model identificerer navngivne værktøjer i tekst uden et lukket kandidatvokabular; output kontrolleres mod tekst og linkes til O*NET eller OOV.
- **Hybrid:** frosne keyword-kandidater verificeres, og modellen kan tilføje værktøjer, der blev overset.
- **HybridVerify:** kun verify/afvis, **ingen tilføjelser**, for at isolere kontekst-check-effekten.

Prompt indeholder regler, eksempler og JSON-schema. I bruger dev-only few-shot-eksempler og en grounding-kontrol, så citeret tekst eller navn skal kunne findes i input. Embedding-linking blev slået fra i den frosne test (`threshold=1.0`), fordi det gav fejlkoblinger på dev.

**Afgørende nuance:** LLM-forsøgene er metodeevaluering, ikke den model som automatisk kører i `app.py`.

## 16. Faktiske testtal: Strict versus lenient for LLM-modeller

*Kilder: `goal1_test_strict.md` og `goal1_test_lenient.md`, samme 30 testopslag.*

| Matcher | Strict precision | Strict recall (alle) | Lenient precision | Lenient recall (alle) |
|---|---:|---:|---:|---:|
| Full keyword | 0,97 | 0,73 | 0,97 | 0,73 |
| LLMOnly | 0,78 | 0,73 | **0,98** | 0,92 |
| Hybrid | 0,77 | 0,75 | 0,97 | **0,95** |
| HybridVerify | **0,98** | 0,73 | **0,98** | 0,73 |

**Hvorfor så stor forskel?** Strict scoring tæller ikke navngivne værktøjer uden for O*NET som korrekte. Hvis LLM'en identificerer et faktisk værktøj uden for vokabularet, straffes den under strict. Lenient scoring tillader match på normaliserede navne mod out-of-vocabulary gold-labels for **alle** metoder. Den er mere egnet til open-vocabulary-sammenligningen, men afhænger af navnenormaliseringen og skal derfor beskrives præcist.

**Baseret på den frosne lenient-test:** Hybrid forbedrer recall (alle) fra 0,73 til 0,95, dvs. **+22 procentpoint**, med et parret 95 %-bootstrapinterval **+0,15 til +0,29**. Precision er begge steder ca. 0,97.

**Nuance:** Hybridens forbedring i *vocabulary-only F1* er lille, og det parrede interval inkluderer 0. Det ville derfor være forkert at sige, at hybrid er statistisk sikkert bedst på *alle* mål.

## 21. Øvelse: Forklar en confusion matrix uden et diagram

I ekstraktion måler I på *værktøj pr. tekst*. Et manuelt label-sæt er ground truth, og en matcher leverer det fundne set.

- **TP**: fundet værktøj der står i gold.
- **FP**: fundet værktøj der ikke står i gold.
- **FN**: gold-værktøj der ikke blev fundet.

Precision = TP/(TP+FP). Recall = TP/(TP+FN). F1 er det harmoniske gennemsnit af precision og recall.

**Hvorfor F1?** Det straffer ubalance mellem de to. Men »recall (vocab)« og »recall (all)« bruger forskellige nævnere og må **ikke** forveksles.

````python
gold = {"Python", "SQL", "Docker", "Databricks"}
found = {"Python", "SQL", "Tableau"}
tp = len(gold & found)
fp = len(found - gold)
fn = len(gold - found)
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1 = 2 * precision * recall / (precision + recall)
print(f"TP={tp}, FP={fp}, FN={fn}")
print(f"Precision={precision:.3f}, Recall={recall:.3f}, F1={f1:.3f}")
assert (tp, fp, fn) == (2,1,2)
````

**Forklaring af output:** To af de fundne tools er korrekte, ét er et falsk positivt fund, og to gold-tools er overset. Det giver precision 2/3, recall 2/4 og F1 ca. 0,571. **Eksamen:** Kan du pege på TP, FP og FN uden at bruge formlerne som krykke?

## 22. Øvelse: Det vigtige skel mellem OOV og almindelige misses

Hvis »Databricks« er relevant men ikke findes i jeres aktuelle O*NET-vokabular, kan keyword-matcheren aldrig finde det via samme vocabulary. Det er ikke nødvendigvis fordi token-algoritmen er dårlig; det er en **designbegrænsning ved kataloget**.

For at vurdere det retfærdigt rapporterer I både `recall (vocab)` og `recall (all)`. Det sidste regner OOV-gold-tools med som missede.

````python
gold = {"Python", "SQL", "Docker", "Databricks"}
vocab = {"Python", "SQL", "Docker", "Tableau"}
found = {"Python", "SQL"}
print("Recall i vocabulary:", len(found & gold & vocab) / len(gold & vocab))
print("Recall på alle værktøjer:", len(found & gold) / len(gold))
assert len(gold - vocab) == 1
````

**Output:** 2/3 = 66,7 % inden for ordforrådet mod 2/4 = 50 % for alle. Den præcise forskel svarer til det vigtige problem i jeres faktiske evaluering: 47 af 204 gold-værktøjsforekomster var uden for vocab.

## 23. Øvelse: Demand og coverage er to forskellige mål

Et værktøjs `share` er **andelen af jobopslag**, der nævner det. Coverage er **summen af de andele, CV'et dækker, divideret med summen af alle valgte topværktøjers andele**. Selv hvis alle tools tilsammen summerer til mere end 100 %, kan coverage godt være et tal mellem 0 og 1.

````python
demand = pd.DataFrame({"tool": ["Python","SQL","Docker","AWS"],
                       "share": [0.75,0.60,0.35,0.20]})
cv_tools = {"Python","SQL"}
demand["on_cv"] = demand["tool"].isin(cv_tools)
coverage = demand.loc[demand.on_cv,"share"].sum() / demand["share"].sum()
print(demand.to_string(index=False))
print(f"Vægtet coverage: {coverage:.1%}")
print("Uvægtet værktøjsandel:", f"{demand.on_cv.mean():.1%}")
````

**Output:** Vægtet coverage er større end 50 %, selvom CV'et har 2 af 4 værktøjer. **Hvorfor?** Python og SQL er de to mest efterspurgte tools i eksemplet. Forklar, hvorfor det er en bedre prioriteringsheuristik end at tælle alle værktøjer lige, men ikke et egentligt mål for ansættelsessandsynlighed.

## 24. Øvelse: Wilson-intervaller bliver bredere med mindre sample

Usikkerheden kommer ikke blot af, at én andel er høj eller lav. Antallet af jobopslag betyder også noget. Den faktiske `demand.py` bruger Wilson score interval med z = 1.96.

````python
def wilson(k, n, z=1.96):
    if n == 0: return (0.0,0.0)
    p=k/n
    denom=1+z*z/n
    centre=(p+z*z/(2*n))/denom
    half=z*math.sqrt(p*(1-p)/n+z*z/(4*n*n))/denom
    return max(0,centre-half), min(1,centre+half)
for n in [20,100,1000]:
    lo,hi=wilson(int(.4*n),n)
    print(f"40 % demand, n={n:4d}: interval {lo:.1%}–{hi:.1%}, bredde {(hi-lo)*100:.1f} pp")
````

**Fortolkning:** To tools kan have samme estimerede andel, men forskellig præcision af estimatet afhængigt af n. Et interval må dog ikke fremstilles som en kur mod scraper-bias: et stort *ikke-repræsentativt* sample kan stadig give misvisende populationspåstande.

## 25. Øvelse: Spearman undersøger ranking, ikke absolutte procenter

Hvis værktøj A er mere efterspurgt end B i begge datasæt, øger det rangkorrelationen. Men hvis niveauerne forskyder sig fra eksempelvis 80 % til 50 %, kan rangeringen stadig være den samme.

````python
a = pd.Series([0.8,0.6,0.4,0.2], index=["Python","SQL","Docker","AWS"])
b = pd.Series([0.5,0.4,0.3,0.1], index=a.index)
c = pd.Series([0.2,0.4,0.6,0.8], index=a.index)
def spearman_demo(x,y):
    return float(np.corrcoef(x.rank(),y.rank())[0,1])
print("Samme rækkefølge:", spearman_demo(a,b))
print("Omvendt rækkefølge:", spearman_demo(a,c))
print("Samtidig er shares forskellige:")
print(pd.DataFrame({"primary":a,"secondary":b}))
````

**Output:** ρ = 1 for samme rækkefølge, selv om andelene ikke er ens, og ρ = −1 for præcis omvendt rækkefølge. Det er derfor, jeres Goal 2-resultater understøtter *replikation af ranking* mere end *lighed i absolut efterspørgsel*.

## 26. Øvelse: Bootstrap-rang bliver usikker ved næsten lige shares

Bootstrap trækker gentagne gange jobopslag **med tilbagelægning** og genberegner rangeringer. Hvis to teknologier ligger tæt, bytter de lettere placering. I den faktiske kode bruger I en posting×tool-matrix og multinomialt fordelte resampling-vægte.

````python
rng=np.random.default_rng(7)
# Syntetisk 0/1-matrix: 100 opslag × 3 værktøjer.
X=np.column_stack([rng.binomial(1,0.70,100),rng.binomial(1,0.51,100),rng.binomial(1,0.49,100)])
W=rng.multinomial(len(X),np.full(len(X),1/len(X)),size=500)
boot_shares=(W@X)/len(X)
boot_ranks=(-boot_shares).argsort(axis=1).argsort(axis=1)+1
for i,name in enumerate(["Python","SQL","Docker"]):
    lo,hi=np.quantile(boot_ranks[:,i],[.025,.975])
    print(name, "share",round(X[:,i].mean(),2),"rank 95 %",(lo,hi))
````

**Fortolkning:** Python ligger oftest stabilt højt, mens SQL og Docker kan bytte rang. Når I præsenterer en top-15-liste, bør I ikke påstå, at nummer 12 nødvendigvis er vigtigere end nummer 13, hvis rankintervallerne overlapper meget.

## 27. Øvelse: Multiple tests og Benjamini–Hochberg

Når I undersøger mange værktøjer i samme junior/senior-analyse, stiger risikoen for tilfældige fund. I `robustness.py` beregner I p-værdier fra en tosidet z-test for forskel på to andele og justerer dem med BH. `q < 0.05` betyder, at resultatet er signifikant efter den valgte multiple-testing-procedure.

````python
def bh_adjust(p):
    p=np.asarray(p,float)
    n=len(p)
    order=np.argsort(p)
    q=p[order]*n/np.arange(1,n+1)
    q=np.minimum.accumulate(q[::-1])[::-1]
    result=np.empty(n)
    result[order]=np.minimum(q,1.0)
    return result
p=np.array([.001,.019,.034,.22,.88])
print(pd.DataFrame({"p":p,"BH-q":bh_adjust(p),"p<0.05":p<.05,"q<0.05":bh_adjust(p)<.05}))
````

**Output:** Nogle rå p-værdier under 0,05 er ikke længere under 0,05 efter BH. Derfor må I ikke hævde, at »enhver forskel med rå p < 0,05 er bevist« efter at have testet mange tools. Husk også: Signifikans er ikke det samme som stor praktisk betydning eller kausalitet.

## 28. Øvelse: Quiz om scorerens definition

Denne øvelse er uafhængig af jeres evalueringsprogram, men demonstrerer, hvorfor strict og lenient scoring kan ændre LLM-resultater dramatisk.

````python
# Syntetisk illustration: Gold inkluderer to OOV-værktøjer.
gold={"Python","SQL","OOV:Databricks","OOV:Jupyter"}
pred={"Python","SQL","OOV:Databricks"}
vocab={"Python","SQL"}
strict_tp=len(pred&gold&vocab)  # i illustrationen krediteres kun vocab
lenient_tp=len(pred&gold)      # OOV krediteres ved navnematch
print("Strict TP (kun in-vocab):",strict_tp)
print("Lenient TP (inkl. korrekt navnematchet OOV):",lenient_tp)
````

**Extractor-begrænsninger:** O*NET-vocab fanger ikke alle tools, og enkeltstående navne kan være tvetydige. En CV-parser kan miste tekst fra layout/scannede dokumenter.

**Score-begrænsninger:** Coverage måler de efterspurgte navngivne tools i CV'et, ikke faktisk erfaring, kravets nødvendighed, jobsandsynlighed eller personlig læringsværdi.

**Evalueringens generaliserbarhed:** 30 hold-out jobopslag for én rolle, en separat lille CV-test på fire dokumenter og begrænset observeret pilotfeedback. Statistisk usikkerhed og udvalgsbias skal fortolkes separat.

**Model-sammenligningens fairness:** Strict og lenient scorer forskellige ting; retfærdig sammenligning kræver eksplicit valg af metric og omgang med OOV.
